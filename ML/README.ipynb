# 1. Descripción General

Este proyecto implementa una solución de Machine Learning sobre Microsoft Fabric para resolver dos problemas de negocio utilizando datos de ventas y clientes:

1. **Clasificación:** Predicción de abandono de clientes (Churn).
2. **Regresión:** Estimación del valor potencial de un cliente.

Todo el flujo fue desarrollado siguiendo las capacidades de Microsoft Fabric, utilizando Lakehouse, MLflow y Model Registry.

---

# 2. Objetivos de Negocio

## Problema 1: Predicción de Churn

### Objetivo

Identificar clientes con alta probabilidad de abandonar la empresa para implementar campañas de retención.

### Impacto Esperado

* Reducir pérdida de clientes.
* Mejorar fidelización.
* Optimizar campañas de marketing.

### Tipo de Problema

Clasificación Binaria.

### Algoritmo Utilizado

Random Forest Classifier.

### Justificación

Random Forest ofrece:

* Buen desempeño con datos tabulares.
* Interpretabilidad aceptable.
* Robustez frente a ruido.
* Bajo riesgo de overfitting.

---

## Problema 2: Valor del Cliente

### Objetivo

Estimar el valor potencial generado por cada cliente.

### Impacto Esperado

* Identificar clientes estratégicos.
* Priorizar inversiones comerciales.
* Mejorar segmentación de mercado.

### Tipo de Problema

Regresión Supervisada.

### Algoritmo Utilizado

Ridge Regression.

### Justificación

* Modelo simple y eficiente.
* Reduce sobreajuste mediante regularización.
* Fácil interpretación.
