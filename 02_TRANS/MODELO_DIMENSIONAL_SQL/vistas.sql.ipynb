## **VISTAS.SQL — Vistas para Consumo Analítico**

Se crean 3 vistas finales sobre el modelo estrella. Estas vistas están diseñadas 
para ser consumidas directamente por la capa de visualización y el modelo semántico en Gold.
Todos los objetos incluyen `DROP VIEW IF EXISTS` para garantizar re-ejecutabilidad.

| Vista | Descripción |
|---|---|
| `vista_ventas_por_categoria` | Ingresos, unidades y transacciones por categoría |
| `vista_ventas_mensuales` | Evolución mensual de ventas |
| `vista_ventas_por_genero` | Comparativo de ventas por género |

## **Vista 1 — vista_ventas_por_categoria**
Consolida el desempeño comercial por categoría de producto uniendo 
`Hechos_Ventas` con `dim_categoria`. Diseñada para alimentar gráficos 
de barras y rankings en el dashboard y el modelo semántico.

| Columna | Descripción |
|---|---|
| `categoria` | Nombre de la categoría de producto |
| `total_transacciones` | Número de ventas realizadas por categoría |
| `total_unidades` | Unidades vendidas por categoría |
| `total_ingresos` | Ingresos totales por categoría en $ ordenados de mayor a menor |

```python
%%sql
DROP VIEW IF EXISTS vista_ventas_por_categoria;

CREATE VIEW vista_ventas_por_categoria AS
SELECT
    dc.product_category                  AS categoria,
    COUNT(h.transaction_id)              AS total_transacciones,
    SUM(h.quantity)                      AS total_unidades,
    ROUND(SUM(h.total_amount), 2)        AS total_ingresos
FROM Hechos_Ventas h
LEFT JOIN dim_categoria dc ON h.id_categoria = dc.id_categoria
GROUP BY dc.product_category
ORDER BY total_ingresos DESC;
```

## **Validación — vista_ventas_por_categoria**
Prueba de ejecución de la vista para confirmar que retorna resultados 
correctos y está disponible para consumo por la capa de visualización 
y el modelo semántico en Gold.

```python
%%sql
SELECT * FROM vista_ventas_por_categoria;
```

## **Vista 2 — vista_ventas_mensuales**
Consolida la evolución temporal de ventas uniendo `Hechos_Ventas` con 
`dim_fecha` para exponer métricas mensuales agregadas. Diseñada para 
alimentar gráficos de tendencia en el dashboard y el modelo semántico.

| Columna | Descripción |
|---|---|
| `anio_mes` | Período en formato `yyyy-MM` para ordenamiento cronológico |
| `total_transacciones` | Número de ventas realizadas en el mes |
| `total_unidades` | Unidades despachadas en el mes |
| `total_ingresos` | Ingresos totales del mes en $ |

```python
%%sql
DROP VIEW IF EXISTS vista_ventas_mensuales;

CREATE VIEW vista_ventas_mensuales AS
SELECT
    DATE_FORMAT(df.fecha, 'yyyy-MM')     AS anio_mes,
    COUNT(h.transaction_id)              AS total_transacciones,
    SUM(h.quantity)                      AS total_unidades,
    ROUND(SUM(h.total_amount), 2)        AS total_ingresos
FROM Hechos_Ventas h
LEFT JOIN dim_fecha df ON h.id_fecha = df.id_fecha
GROUP BY DATE_FORMAT(df.fecha, 'yyyy-MM')
ORDER BY anio_mes;
```

## **Validación — vista_ventas_mensuales**
Prueba de ejecución de la vista para confirmar que retorna resultados 
correctos y está disponible para consumo por la capa de visualización 
y el modelo semántico en Gold.

```python
%%sql
SELECT * FROM vista_ventas_mensuales;
```

## **Vista 3 — vista_ventas_por_genero**
Consolida el desempeño comercial por género de cliente uniendo 
`Hechos_Ventas` con `dim_genero`. Diseñada para alimentar análisis 
de segmentación demográfica en el dashboard y el modelo semántico.

| Columna | Descripción |
|---|---|
| `genero` | Nombre del género del cliente |
| `total_transacciones` | Número de ventas realizadas por género |
| `total_unidades` | Unidades vendidas por género |
| `total_ingresos` | Ingresos totales por género en $ ordenados de mayor a menor |

```python
%%sql
DROP VIEW IF EXISTS vista_ventas_por_genero;

CREATE VIEW vista_ventas_por_genero AS
SELECT
    dg.nombre_genero                     AS genero,
    COUNT(h.transaction_id)              AS total_transacciones,
    SUM(h.quantity)                      AS total_unidades,
    ROUND(SUM(h.total_amount), 2)        AS total_ingresos
FROM Hechos_Ventas h
LEFT JOIN dim_genero dg ON h.id_genero = dg.id_genero
GROUP BY dg.nombre_genero
ORDER BY total_ingresos DESC;
```
