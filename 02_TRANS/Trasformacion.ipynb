```python
# Welcome to your new notebook
# Type here in the cell editor to add code!
from pyspark.sql.types import *
silver = spark.read.table(
    "bronze.dbo.ventas_raw"
)
display(silver)
```

```python
%%sql

DROP TABLE IF EXISTS silver;

CREATE TABLE silver
AS
SELECT *
FROM bronze.dbo.ventas_raw;
```

```python
%%sql

SELECT *

FROM silver
```

```python
%%sql

-- 1.1 Conteo total de registros en Bronze
SELECT COUNT(*) AS total_registros_silver
FROM silver;
```

```python
%%sql
-- 1.2 Diagnóstico de nulos por columna
-- Identifica cuántos nulos existen en cada campo antes de tratarlos.
SELECT
    COUNT(*) AS total_filas,
    SUM(CASE WHEN transaction_id    IS NULL THEN 1 ELSE 0 END) AS nulos_transaction_id,
    SUM(CASE WHEN date              IS NULL THEN 1 ELSE 0 END) AS nulos_date,
    SUM(CASE WHEN customer_id       IS NULL THEN 1 ELSE 0 END) AS nulos_customer_id,
    SUM(CASE WHEN gender            IS NULL THEN 1 ELSE 0 END) AS nulos_gender,
    SUM(CASE WHEN age               IS NULL THEN 1 ELSE 0 END) AS nulos_age,
    SUM(CASE WHEN product_category  IS NULL THEN 1 ELSE 0 END) AS nulos_product_category,
    SUM(CASE WHEN quantity          IS NULL THEN 1 ELSE 0 END) AS nulos_quantity,
    SUM(CASE WHEN price_per_unit    IS NULL THEN 1 ELSE 0 END) AS nulos_price_per_unit,
    SUM(CASE WHEN total_amount      IS NULL THEN 1 ELSE 0 END) AS nulos_total_amount
FROM silver;
```

```python
%%sql
-- 1.3 Diagnóstico de duplicados en la clave primaria (transaction_id)
-- Un transaction_id duplicado indica ingestión errónea o datos corruptos.
SELECT
    transaction_id,
    COUNT(*) AS cantidad
FROM silver
GROUP BY transaction_id
HAVING COUNT(*) > 1
ORDER BY cantidad DESC;
```
